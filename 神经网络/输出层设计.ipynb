```python
import numpy as np
```

```python
# 恒等函数和softmax函数
# 恒等函数：输出就是输入
```

```python
# softmax函数
```

$$
\text{softmax}(a_i) = \frac{e^{a_i}}{\sum_{j=1}^{K} e^{a_j}}
$$

```python
# k表示最后一层输出层有多少个神经元
# softmax函数的实现
def softmax(a):
    exp_a = np.exp(a)
    sum_exp_a = np.sum(exp_a)
    y = exp_a / sum_exp_a
    return y
```

```python
# 在计算机进行运算的时候可能会出现溢出，需要将softmax函数化简
```

$$
\begin{aligned}
y_k
&= \frac{\exp(a_k)}{\sum_{i=1}^{n} \exp(a_i)} \\
&= \frac{C\exp(a_k)}{C\sum_{i=1}^{n} \exp(a_i)} \\
&= \frac{\exp(a_k + \log C)}{\sum_{i=1}^{n} \exp(a_i + \log C)} \\
&= \frac{\exp(a_k + C')}{\sum_{i=1}^{n} \exp(a_i + C')}
\end{aligned}
$$

```python
# 改进后的softmax函数
def softmax(a):
    c = np.max(a)
    exp_a = np.exp(c-a)
    sum_exp_a = np.sum(exp_a)
    y = exp_a/sum_exp_a
    return y
```
